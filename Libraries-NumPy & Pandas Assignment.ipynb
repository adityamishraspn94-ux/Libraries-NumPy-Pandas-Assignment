Libraries-NumPy & Pandas Assignment

Q1. Load the given retail sales dataset into a Pandas DataFrame and display the first 5 rows along with the dataset information to understand its structure.

```python
import pandas as pd

# load dataset
df = pd.read_csv("retail_dataset .csv")

# first 5 rows
df.head()

# dataset info
df.info()
```

Q2. Inspect the dataset and identify missing values in each column. Display the total number of missing values present in the dataset. 


```python
# missing values count in each column
df.isnull().sum()

# total missing values
df.isnull().sum().sum()
```

Q3. Replace missing values in the Quantity and Price columns with their respective mean values using
Pandas.

```python
# fill missing Quantity with mean
df["Quantity"].fillna(df["Quantity"].mean(), inplace=True)

# fill missing Price with mean
df["Price"].fillna(df["Price"].mean(), inplace=True)
df.head()
```

Q4. Remove all rows where the Product Category or Region column contains missing values.

```python
# drop rows where Product Category or Region is missing
df.dropna(subset=["Product Category", "Region"], inplace=True)
df.head()
```

Q5. Create a new column called Revenue using NumPy, where:

Revenue = Quantity × Price

Display the updated dataset with the new column.

```python
# create new column Revenue
df["Revenue"] = df["Quantity"] * df["Price"]

# check updated data
df.head()
```

Q6. Using NumPy, calculate the total revenue generated from all transactions in the dataset.

```python
# total revenue of all data
df["Revenue"].sum()
```

Q7. Group the dataset by Product Category and calculate the total revenue generated by each category.

```python
# total revenue per category
df.groupby("Product Category")["Revenue"].sum()
```

Q8. Based on the revenue calculated for each product category, identify the top 3 categories with the
highest revenue and the bottom 3 categories with the lowest revenue.

```python
# group data
rev_cat = df.groupby("Product Category")["Revenue"].sum()

# top 3 categories
rev_cat.sort_values(ascending=False).head(3)

# bottom 3 categories
rev_cat.sort_values().head(3)
```

Q9. Group the dataset by Region and calculate the total revenue generated in each region. Identify the
region with the highest revenue and the region with the lowest revenue.

```python
# revenue per region
rev_region = df.groupby("Region")["Revenue"].sum()

# highest revenue region
rev_region.idxmax()

# lowest revenue region
rev_region.idxmin()
```

Q10. Extract the month from the Date column using Pandas and group the dataset by month to calculate the total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column using NumPy.

```python
# convert date (fix applied)
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

# create Month-Year
df["Month_Year"] = df["Date"].dt.strftime("%Y-%m")

# ensure revenue exists
df["Revenue"] = df["Quantity"] * df["Price"]

# monthly revenue
df.groupby("Month_Year")["Revenue"].agg(['sum','mean','median','std'])
```
